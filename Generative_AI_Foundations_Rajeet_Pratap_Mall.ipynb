# Module 1: Generative AI Fundamentals

## 1. Generative AI
Generative AI is a type of AI that learns patterns
from data and generates new content such as text,
images, audio, video and code.

## 2. Large Language Models
LLMs are neural networks trained on large amounts
of language data. They generate text by predicting
the next token based on the preceding context.

## 3. Tokens
Tokens are the units of text processed by a model.
They can represent words, parts of words or symbols.

## 4. Context Window
The context window is the maximum amount of
information a model can process in one interaction.
It is measured in tokens.

## 5. Transformers
Transformers use self-attention to learn relationships
between tokens. Their major components include
embeddings, attention mechanisms and feed-forward
neural network layers.

## Key Learning
Generative AI generates new content. LLMs process
text as tokens, work within context-window limits
and often use the Transformer architecture.

## Module 3: GenAI Application Architecture

A Generative AI application connects a user interface to an AI model and presents the model's response to the user.

### Basic workflow
1. User enters a prompt.
2. The application receives and prepares the prompt.
3. The prompt is sent to an AI model.
4. The model generates a response.
5. The application may validate or format the response.
6. The response is displayed to the user.

### Main components
- Frontend: The user-facing interface.
- Backend: Handles application logic and model requests.
- Model/API: Generates content based on the prompt.
- Knowledge source: Optional documents or databases used to provide context.
- Safety and evaluation: Helps assess output quality, safety and reliability.

### Retrieval-Augmented Generation (RAG)
RAG retrieves relevant information from an external knowledge source and includes it in the model's context. It can make responses more grounded in supplied information, but the output still needs checking.

### Key takeaway
A GenAI application is more than a model alone. It includes the interface, request handling, context, model, output processing and evaluation.

````python
def simple_genai_app(user_prompt):
    prompt = user_prompt.strip()

    if not prompt:
        return "Please enter a prompt."

    # Placeholder for sending the prompt to an AI model
    response = f"Your prompt is ready to be sent to an AI model: {prompt}"

    return response

user_input = "Explain generative AI in one sentence."
print(simple_genai_app(user_input))
````

### Observation
The function accepts a user prompt, checks whether it is empty, and returns a prepared message. This demonstrates the input-handling stage of a GenAI application. It does not generate an AI response because it is not connected to a model.

### Conclusion
A GenAI application needs a workflow that handles user input and model communication. Additional components such as retrieval, output checks and safety evaluation can be added depending on the application's purpose.

# Module 2: Prompt Engineering & AI Productivity


### What is a prompt?
A prompt is an instruction or input provided to an AI model. Prompt engineering is the process of designing and improving prompts to guide the model toward useful outputs.

### Five parts of a structured prompt
1. Role: The perspective or expertise the AI should adopt.
2. Context: Background information relevant to the task.
3. Instruction: The action the AI should perform.
4. Constraints: Rules such as length, tone, audience, or content limits.
5. Output format: The structure in which the answer should be returned.

### Example
Vague prompt: "Write an email."

Structured prompt:
- Role: Professional communication assistant.
- Context: A college student needs an assignment extension due to an exam.
- Instruction: Write a polite email to the professor requesting a two-day extension.
- Constraints: Under 120 words, simple formal English, no invented personal details.
- Output format: Subject line and email body.

### Key takeaway
Clear prompts provide the model with direction and constraints. They can improve relevance and consistency, but generated content should still be checked before use.


## Zero-Shot Prompting

Zero-shot prompting asks an AI model to perform a task without providing examples.

Example:
"Classify the following message as Positive, Negative, or Neutral:
I really enjoyed using this application."

## Few-Shot Prompting

Few-shot prompting provides a small number of examples before asking the model to perform a new task.

Example:

Example 1:
"I love this product."
Classification: Positive

Example 2:
"The application keeps crashing."
Classification: Negative

Example 3:
"The product arrived yesterday."
Classification: Neutral

Task:
"The application is very easy to use."

## Difference

Zero-shot prompting provides instructions without examples.

Few-shot prompting provides instructions along with examples that demonstrate the expected behavior.

## Key Takeaway

Few-shot examples can help guide the model toward a desired pattern or format, especially when the task or expected output is not obvious.

## Zero-Shot vs Few-Shot Experiment

| Technique | Examples Provided | Task | Output |
|---|---|---|---|
| Zero-Shot | No | Sentiment classification | [Negative] |
| Few-Shot | Yes | Sentiment classification | [Negative] |

### Observation

The zero-shot prompt provided only the task instruction, while the few-shot prompt provided examples of the expected classifications. The examples helped demonstrate the required task and output pattern.

### Conclusion

Zero-shot prompting is useful when the task is straightforward and can be clearly described with instructions. Few-shot prompting is useful when examples can help the model understand the expected pattern or format. The choice depends on the complexity and requirements of the task.

#  AI-Assisted Coding and Debugging

## Objective

In this section, I will learn how AI can assist with coding and debugging. I will provide a buggy Python program to an AI model, identify the error, generate a corrected version, and test the solution.

````python
numbers = [80, 70, 90]

total = sum(numbers)
average = total / len(numbers)

print("Average:", average)
````

### Error Observed

The program produced a `NameError` because `number` was used instead of the defined variable `numbers`.

This error is intentional and will be analyzed and corrected using an AI debugging prompt.

## 2. AI-Assisted Debugging

### Debugging Prompt

You are a Python debugging assistant.

Analyze the following code:

numbers = [80, 70, 90]

total = sum(numbers)
average = total / len(number)

print("Average:", average)

Identify the error, explain why it occurs, provide the corrected code, and briefly explain the correction.

## 2. AI-Assisted Debugging

### Debugging Prompt

You are a Python debugging assistant.

Identify the error, explain why it occurs, provide the corrected code, and briefly explain the correction.

````python
numbers = [80, 70, 90]

total = sum(numbers)
average = total / len(numbers)

print("Average:", average)
````

````python
def calculate_average(numbers):
    if not numbers:
        return None

    return sum(numbers) / len(numbers)


# Test 1
print("Test 1:", calculate_average([80, 70, 90]))

# Test 2
print("Test 2:", calculate_average([100, 90]))

# Test 3
print("Test 3:", calculate_average([]))
````

## 3. Testing Report

| Test Case | Input | Expected Result | Actual Result | Status |
|---|---|---|---|---|
| Test 1 | [80, 70, 90] | 80.0 | 80.0 | Pass |
| Test 2 | [100, 90] | 95.0 | 95.0 | Pass |
| Test 3 | [] | None | None | Pass |

### Observation

The original program produced a NameError because it referred to `number` instead of `numbers`. The AI identified the variable-name error and suggested the correction. After fixing the code, the program produced the expected results. Additional test cases were used to check normal inputs and an empty list.

### Conclusion

AI can assist developers by identifying coding errors, explaining why they occur, and suggesting corrections. However, the generated solution should be tested rather than accepted without verification. Testing different inputs helps confirm that the corrected code behaves as expected.
